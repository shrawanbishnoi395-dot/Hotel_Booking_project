```python
import pandas as pd
import numpy as np

pd.set_option("display.max_columns", None)

# Load Dataset
df = pd.read_csv("/content/hotel_bookings.csv.zip")

print(df.shape)

# Working Copy
df_clean = df.copy()
```

```python
duplicate_count = df_clean.duplicated().sum()

print("Duplicate Rows :", duplicate_count)

df_clean.drop_duplicates(inplace=True)

print(df_clean.shape)
```

```python
missing = pd.DataFrame({
    "Missing Values": df_clean.isnull().sum(),
    "Percentage": round(df_clean.isnull().mean()*100,2)
})

missing = missing.sort_values(
    by="Percentage",
    ascending=False
)

missing
```

```python
# children -> median
df_clean["children"] = df_clean["children"].fillna(
    df_clean["children"].median()
)

# country -> Unknown
df_clean["country"] = df_clean["country"].fillna("Unknown")

# agent -> 0
df_clean["agent"] = df_clean["agent"].fillna(0)

# company -> 0
df_clean["company"] = df_clean["company"].fillna(0)
```

```python
df_clean.isnull().sum()
```

```python
df_clean["reservation_status_date"] = pd.to_datetime(
    df_clean["reservation_status_date"]
)
```

```python
# Total Guests

df_clean["total_guests"] = (
    df_clean["adults"]
    + df_clean["children"]
    + df_clean["babies"]
)

# Total Nights

df_clean["total_nights"] = (
    df_clean["stays_in_weekend_nights"]
    + df_clean["stays_in_week_nights"]
)
```

```python
print(df_clean.shape)

print(df_clean.info())
```

```python
df_clean.to_csv(
    "cleaned_hotel_bookings.csv",
    index=False
)

print("Dataset Saved Successfully ✅")
```
